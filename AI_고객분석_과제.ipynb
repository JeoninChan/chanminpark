# AI사무업무활용 데이터 분석 과제

고객 데이터 200명을 기반으로 한 분석 및 시각화 실습입니다.

```python
import pandas as pd

# 데이터 불러오기
df = pd.read_csv('/content/가공_고객_데이터_200명_UTF8BOM.csv', encoding='utf-8-sig')
df.head()
```

## 1. 기초 분석

```python
df.columns
```

```python
df['성별'].value_counts()
```

```python
df.groupby('회원등급')['구매금액(만원)'].mean()
```

```python
df['나이'].describe()[['mean', 'min', 'max']]
```

## 2. 시각화

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.rcParams['font.family'] = 'AppleGothic' if 'AppleGothic' in plt.rcParams['font.family'] else 'NanumGothic'
sns.set(style='whitegrid')
```

```python
sns.barplot(data=df, x='성별', y='구매금액(만원)', estimator='mean')
plt.title('성별별 평균 구매금액')
plt.show()
```

```python
plt.hist(df['나이'], bins=10, color='skyblue', edgecolor='black')
plt.title('고객 나이 분포')
plt.xlabel('나이')
plt.ylabel('고객 수')
plt.show()
```

```python
sns.boxplot(data=df, x='회원등급', y='구매금액(만원)')
plt.title('회원등급별 구매금액 분포')
plt.show()
```

## 3. 관계 분석

```python
correlation = df[['나이', '구매금액(만원)']].corr()
print('나이와 구매금액 상관계수:\n', correlation)
```

```python
sns.scatterplot(data=df, x='나이', y='구매금액(만원)', hue='회원등급')
plt.title('나이 vs 구매금액 산점도')
plt.show()
```
